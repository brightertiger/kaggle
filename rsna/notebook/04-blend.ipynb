```python
import pandas as pd
import numpy as np
```

```python
data_1 = pd.read_csv('../../score/score_6.csv')
data_2 = pd.read_csv('../../score/score_7.csv')
data_3 = pd.read_csv('../../score/score_8.csv')
data_4 = pd.read_csv('../../score/score_9.csv')
#data_5 = pd.read_csv('../../score/score_5.csv')
```

```python
data = data_1.append(data_2).append(data_3).append(data_4)
data = data.groupby('image').mean().reset_index() 
```

```python
data = pd.melt(data, id_vars='image')
```

```python
data['ID'] = 'ID_' + data['image'] + '_' + data['variable']
data['Label'] = data['value']
data = data[['ID','Label']]
```

```python
data.to_csv('submit.csv', index=False)
```

```python
import pydicom
import pandas as pd
import numpy as np
```

```python
features = ['SOPInstanceUID','StudyID','StudyInstanceUID','PatientID','SeriesInstanceUID']
features += ['BitsAllocated','BitsStored','Columns','HighBit','PixelRepresentation','Rows','SamplesPerPixel']
features += ['RescaleIntercept','RescaleSlope','Modality','PhotometricInterpretation']
features += ['ImagePositionPatient','ImageOrientationPatient','PixelSpacing']
```

```python
def processDicom(dicom):
    global features
    data = {}
    try:
        for column in features:
            data[column] = str(getattr(dicom, column))
    except:
        print('error')
    return data
```

```python
train = pd.read_csv('../data/train.csv')
train['ID'] = train['ID'].map(lambda x : '_'.join(x.split('_')[:-1])) 
train = train['ID'].unique().tolist()
```

```python
data = []
for idx, name in enumerate(train):
    if idx % 10000 == 0:
        print(idx, 'images processed...')
    dicom = pydicom.dcmread('../data/train/{}.dcm'.format(name))
    meta = processDicom(dicom)
    data.append(meta)
```

```python
data = pd.DataFrame.from_records(data)
data = data[features]
```

```python
data.to_csv('../data/train_meta.csv', index=False)
```

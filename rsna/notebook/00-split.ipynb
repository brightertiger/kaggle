```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
```

```python
meta = pd.read_csv('../../data/train_meta.csv')
meta = meta[['SOPInstanceUID','PatientID']]
meta.columns = ['image','patient']
```

```python
folds = meta[['patient']].drop_duplicates().reset_index(drop=True)
folds['fold'] = 0
split = KFold(n_splits=5, random_state=2017)
counter = 1
for train_idx, valid_idx in split.split(folds):
    folds.loc[valid_idx, 'fold'] = counter
    counter += 1
folds['fold'].value_counts()
```

```python
meta = meta.merge(folds, on=['patient'])
meta['image'] = meta['image'].map(lambda x : x.split('_')[1])
meta['fold'].value_counts()
```

```python
data = pd.read_csv('../../data/stage_1_train.csv')
```

```python
data['image'] = data['ID'].map(lambda x : x.split('_')[1])
data['class'] = data['ID'].map(lambda x : x.split('_')[-1])
data = data.pivot_table(index='image',columns='class',values='Label', aggfunc='sum')
data = data.reset_index()
data = meta.merge(data, on='image')
```

```python
data = data[['image','fold','any','epidural','intraparenchymal','intraventricular','subarachnoid','subdural']]
```

```python
data.to_csv('../../data/train.csv', index=False)
```

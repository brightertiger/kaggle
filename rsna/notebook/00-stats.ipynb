```python
import pandas as pd
import numpy as np
```

```python
label = pd.read_csv('../../data/train.csv')
```

```python
stats = pd.read_csv('../../data/train_stats.csv')
stats['image'] = stats['image'].map(lambda x : x.replace('ID_',''))
```

```python
data = label.merge(stats, on='image')
```

```python
data.shape
```

```python
data[data['percent'] <= 0.01]['any'].agg(['sum','count'])
```

```python
ignore = data[data['percent'] <= 0.01]['image'].tolist()
```

```python
label = label[~label['image'].isin(ignore)]
```

```python
label.to_csv('../../data/train.csv', index=False)
```

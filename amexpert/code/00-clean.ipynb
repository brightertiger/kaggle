```python
import pandas as pd
import numpy as np
```

```python
def clean(x):
    day = x.split('/')[0]
    month = x.split('/')[1]
    year = x.split('/')[2]
    return '20' + year + '-' + month + '-' + day
```

```python
campaign_data = pd.read_csv('../data/data/campaign_data.csv', parse_dates=False)
```

```python
campaign_data.head()
```

```python
campaign_data['start_date'] = campaign_data['start_date'].map(clean)
campaign_data['end_date'] = campaign_data['end_date'].map(clean)
```

```python
campaign_data.head()
```

```python
campaign_data.to_csv('../data/data/campaign_data_clean.csv', index=False)
```

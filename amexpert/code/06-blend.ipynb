```python
import pandas as pd
import numpy as np
```

```python
score_1 = pd.read_csv('../data/score/score_v1.csv')
score_2 = pd.read_csv('../data/score/score_v2.csv')
score_3 = pd.read_csv('../data/score/score_v3.csv')
```

```python
pd.Series.corr(score_1.redemption_status, score_2.redemption_status)
```

```python
pd.Series.corr(score_1.redemption_status, score_3.redemption_status)
```

```python
score_1.redemption_status.mean(), score_2.redemption_status.mean(), score_3.redemption_status.mean()
```

```python
score_1['redemption_status'] = score_1['redemption_status'].rank()
score_2['redemption_status'] = score_2['redemption_status'].rank()
score_3['redemption_status'] = score_3['redemption_status'].rank()
```

```python
score = score_1.append(score_2).append(score_3)
score = score.groupby('id').mean().reset_index()
```

```python
score.to_csv('score.csv', index=False)
```

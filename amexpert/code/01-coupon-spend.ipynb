```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('../data/data/customer_transaction_data.csv')
mapping = pd.read_csv('../data/data/coupon_item_mapping.csv')
data = data.merge(mapping, on='item_id')
```

```python
data.head()
```

```python
feat_1 = data.groupby('coupon_id')['selling_price'].sum().reset_index()
feat_2 = data.groupby('coupon_id')['quantity'].sum().reset_index()
feat_3 = data.groupby('coupon_id')['other_discount'].sum().reset_index()
feat_4 = data.groupby('coupon_id')['coupon_discount'].sum().reset_index()
feat_5 = data.groupby('coupon_id')['coupon_discount'].agg(np.count_nonzero).reset_index()
```

```python
feat_1.columns = ['coupon_id','sum_coup_prc'] 
feat_2.columns = ['coupon_id','sum_coup_qty'] 
feat_3.columns = ['coupon_id','sum_coup_dsc']
feat_4.columns = ['coupon_id','sum_coup_cdsc']
feat_5.columns = ['coupon_id','cnt_coup_cdsc']
```

```python
driver = pd.read_csv('../data/driver.csv')
```

```python
driver = driver.merge(feat_1, on='coupon_id')
driver = driver.merge(feat_2, on='coupon_id')
driver = driver.merge(feat_3, on='coupon_id')
driver = driver.merge(feat_4, on='coupon_id')
driver = driver.merge(feat_5, on='coupon_id')
```

```python
driver = driver.drop(['campaign_id','coupon_id','customer_id'], axis=1)
```

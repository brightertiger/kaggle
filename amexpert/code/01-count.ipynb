```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('../data/driver.csv')
```

```python
data['cnt_cust'] = data.groupby('customer_id')['id'].transform('count')
data['cnt_camp'] = data.groupby('campaign_id')['id'].transform('count')
data['cnt_coup'] = data.groupby('coupon_id')['id'].transform('count')
data['cnt_cust_coup'] = data.groupby(['coupon_id','customer_id'])['id'].transform('count')
data['cnt_camp_coup'] = data.groupby(['campaign_id','coupon_id'])['id'].transform('count')
data['cnt_camp_cust'] = data.groupby(['campaign_id','customer_id'])['id'].transform('count')
```

```python
data['dsc_cust_coup'] = data.groupby('customer_id')['coupon_id'].transform(pd.Series.nunique)
data['dsc_cust_camp'] = data.groupby('customer_id')['campaign_id'].transform(pd.Series.nunique)
data['dsc_camp_coup'] = data.groupby('campaign_id')['coupon_id'].transform(pd.Series.nunique)
data['dsc_coup_camp'] = data.groupby('coupon_id')['campaign_id'].transform(pd.Series.nunique)
```

```python
data = data.drop(['campaign_id','coupon_id','customer_id'], axis=1)
```

```python
data.to_csv('../data/feature/count_feature.csv', index=False)
```

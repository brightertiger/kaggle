```python
import pandas as pd
import numpy as np
```

```python
subset = [13]
```

```python
driver = pd.read_csv('../data/driver.csv').drop(['coupon_id','campaign_id'], axis=1)
train = pd.read_csv('../data/data/train.csv')
valid = train[train['campaign_id'].isin(subset)][['id','redemption_status']]
train = train[~train['campaign_id'].isin(subset)][['id','redemption_status']]
test = pd.read_csv('../data/data/test.csv')[['id']]
print(driver.shape, train.shape, valid.shape, test.shape)
```

```python
data_1 = pd.read_csv('../data/feature/customer_feature.csv')
data_2 = pd.read_csv('../data/feature/campaign_feature.csv')
data_3 = pd.read_csv('../data/feature/coupon_feature.csv')
data_4 = pd.read_csv('../data/feature/coupon_spend_profile.csv')
data_5 = pd.read_csv('../data/feature/count_feature.csv')
data_6 = pd.read_csv('../data/feature/tranx_brand_feature.csv')
data_7 = pd.read_csv('../data/feature/tranx_category_feature.csv')
data_8 = pd.read_csv('../data/feature/tranx_item_feature.csv')
data_9 = pd.read_csv('../data/feature/similarity.csv')
data_10 = pd.read_csv('../data/feature/tranx_time_feature.csv')
```

```python
driver = driver.merge(data_1, on='id')
driver = driver.merge(data_2, on='id')
driver = driver.merge(data_3, on='id')
driver = driver.merge(data_4, on='id')
driver = driver.merge(data_5, on='id')
driver = driver.merge(data_6, on='id')
driver = driver.merge(data_7, on='id')
driver = driver.merge(data_8, on='id')
driver = driver.merge(data_9, on='id')
driver = driver.merge(data_10, on='id')
```

```python
train = train.merge(driver, on='id')
valid = valid.merge(driver, on='id')
test = test.merge(driver, on='id')
print(driver.shape, train.shape, valid.shape, test.shape)
```

```python
train.to_csv('../data/model/train.csv', index=False)
valid.to_csv('../data/model/valid.csv', index=False)
test.to_csv('../data/model/test.csv', index=False)
```

```python
import pandas as pd
import numpy as np
```

```python
train = pd.read_csv('../data/data/train.csv')
test = pd.read_csv('../data/data/test.csv')
test['redemption_status'] = 0
```

```python
def calcOverlap(feature):
    train_feat = set(train[feature])
    test_feat = set(test[feature])
    print('Feature:', feature)
    print('Train:', len(train_feat))
    print('Test:', len(test_feat))
    print('Overlap:', len(train_feat & test_feat))
    return None
```

```python
calcOverlap('campaign_id')
calcOverlap('coupon_id')
calcOverlap('customer_id')
```

```python
campaign = pd.read_csv('../data/data/campaign_data_clean.csv')
campaign['start_date'] = pd.to_datetime(campaign['start_date'])
campaign['end_date'] = pd.to_datetime(campaign['end_date'])
metrics = train.append(test)
metrics = metrics.groupby(['campaign_id'])['customer_id'].count().reset_index()
metrics.columns = ['campaign_id','volume']
campaign = campaign.merge(metrics, on='campaign_id').reset_index(drop=True)
campaign = campaign.sort_values(by='start_date')
lookup = train['campaign_id'].unique()
mapping = lambda x : 'train' if x in lookup else 'test'
campaign['split'] = campaign['campaign_id'].map(lambda x : mapping(x))
```

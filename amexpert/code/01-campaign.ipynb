```python
import pandas as pd
import numpy as np
```

```python
driver = pd.read_csv('../data/driver.csv')[['id','campaign_id']]
```

```python
data = pd.read_csv('../data/data/campaign_data_clean.csv')
```

```python
data['campaign_type'] = data['campaign_type'].map(lambda x : 1 if x == 'X' else 0)
```

```python
data['start_date'] = pd.to_datetime(data['start_date']) 
data['end_date'] = pd.to_datetime(data['end_date'])
```

```python
data['day'] = data['start_date'].dt.day
data['month'] = data['start_date'].dt.month
data['duration'] = (data['end_date'] - data['start_date']).dt.days
```

```python
data = data.drop(['end_date','start_date'], axis=1)
data = driver.merge(data, on='campaign_id', how='left').drop('campaign_id', axis=1)
```

```python
data.to_csv('../data/feature/campaign_feature.csv', index=False)
```

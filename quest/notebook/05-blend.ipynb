```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr
```

```python
def evaluateModel(labels, scores):
    correls = []
    for idx in range(30):
        label = labels.iloc[:,idx]
        score = scores.iloc[:,idx]
        print(score.nunique())
        correl = spearmanr(label, score).correlation
        correls.append(correl)
    metric = round(np.mean(correls), 4)
    print('Metric:', metric)
    return None
```

```python
labels = pd.read_csv('../../model/version-2/labels.csv')
```

```python
score_1 = pd.read_csv('../../model/version-2/scores.csv').reset_index()
score_2 = pd.read_csv('../../model/version-2/scores.csv').reset_index()
```

```python
scores = score_1.append(score_2)
scores = scores.groupby('index').mean().reset_index()
scores.iloc[:,1:] = (scores.iloc[:,1:] * 18).round(0)
scores.iloc[:,1:] = scores.iloc[:,1:] / 36
```

```python
evaluateModel(labels, scores.iloc[:,1:])
```

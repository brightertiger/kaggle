```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import GroupKFold
```

```python
LABELS =  ['question_asker_intent_understanding','question_body_critical', 'question_conversational']
LABELS += ['question_expect_short_answer', 'question_fact_seeking', 'question_has_commonly_accepted_answer']
LABELS += ['question_interestingness_others', 'question_interestingness_self', 'question_multi_intent']
LABELS += ['question_not_really_a_question', 'question_opinion_seeking', 'question_type_choice']
LABELS += ['question_type_compare', 'question_type_consequence', 'question_type_definition']
LABELS += ['question_type_entity', 'question_type_instructions','question_type_procedure']
LABELS += ['question_type_reason_explanation', 'question_type_spelling', 'question_well_written']
LABELS += ['answer_helpful', 'answer_level_of_information', 'answer_plausible']
LABELS += ['answer_relevance', 'answer_satisfaction', 'answer_type_instructions']
LABELS += ['answer_type_procedure','answer_type_reason_explanation','answer_well_written']
```

```python
TEXT = ['question_title','question_body','answer']
USERS = ['question_user_name','answer_user_name']
URLS = ['url','question_user_page','answer_user_page']
CATS = ['category','host']
```

```python
data = pd.read_csv('../../data/download/train.csv')
data = data.sample(frac=1., random_state=2017)
data = data.reset_index(drop=True)
```

```python
folds = GroupKFold(n_splits=5).split(X=data.qa_id, groups=data.question_body)
```

```python
idx = 0
for train_idx, valid_idx in folds:
    idx += 1
    train = data.iloc[train_idx, :].reset_index(drop=True)
    valid = data.iloc[valid_idx, :].reset_index(drop=True)
    print('Data:', train.shape, valid.shape)
    train[['qa_id'] + TEXT].to_csv('../../data/split/text_train_{}.csv'.format(idx), index=False)
    valid[['qa_id'] + TEXT].to_csv('../../data/split/text_valid_{}.csv'.format(idx), index=False)
    train[['qa_id'] + USERS + URLS + CATS].to_csv('../../data/split/data_train_{}.csv'.format(idx), index=False)
    valid[['qa_id'] + USERS + URLS + CATS].to_csv('../../data/split/data_valid_{}.csv'.format(idx), index=False)
    train[['qa_id'] + LABELS].to_csv('../../data/split/label_train_{}.csv'.format(idx), index=False)
    valid[['qa_id'] + LABELS].to_csv('../../data/split/label_valid_{}.csv'.format(idx), index=False)
```

```python
score = pd.read_csv('../../data/download/test.csv')
score[['qa_id'] + TEXT].to_csv('../../data/split/text_score.csv', index=False)
score[['qa_id'] + USERS + URLS + CATS].to_csv('../../data/split/data_score.csv', index=False)
```

```python
import pandas as pd
import numpy as np
import soundfile as sf
import librosa as lb
```

```python
data = pd.read_csv('../../data/positive.csv')
data = data['recording_id'].unique().tolist()
```

```python
for name in data:
    sound, orig_sr = sf.read('../../data/raw/train/{}.flac'.format(name))
    sound = lb.resample(sound, orig_sr=orig_sr, target_sr=32000, res_type="kaiser_best")
    np.save('../../data/resample/train/{}.npy'.format(name), sound)
```

```python
data = pd.read_csv('../../data/raw/sample_submission.csv')
data = data['recording_id'].unique().tolist()
```

```python
for name in data:
    sound, orig_sr = sf.read('../../data/raw/test/{}.flac'.format(name))
    sound = lb.resample(sound, orig_sr=orig_sr, target_sr=32000, res_type="kaiser_best")
    np.save('../../data/resample/test/{}.npy'.format(name), sound)
```

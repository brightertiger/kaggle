```python
import pandas as pd
import numpy as np
```

```python
score0 = pd.read_csv('../../score/version2.csv')
score1 = pd.read_csv('../../score/version1.csv') 
score2 = pd.read_csv('../../score/version5.csv')
score3 = pd.read_csv('../../score/version3.csv')
score4 = pd.read_csv('../../score/version0.csv')
score5 = pd.read_csv('../../score/public_857.csv')
score6 = pd.read_csv('../../score/public_880.csv')
score0.iloc[:,1:] = score0.iloc[:,1:].rank(axis=1)
score1.iloc[:,1:] = score1.iloc[:,1:].rank(axis=1)
score2.iloc[:,1:] = score2.iloc[:,1:].rank(axis=1)
score3.iloc[:,1:] = score3.iloc[:,1:].rank(axis=1)
score4.iloc[:,1:] = score4.iloc[:,1:].rank(axis=1)
score5.iloc[:,1:] = score5.iloc[:,1:].rank(axis=1)
score6.iloc[:,1:] = score6.iloc[:,1:].rank(axis=1)
```

```python
score = score0.append(score1).append(score2).append(score3).append(score4).append(score5).append(score6)
score = score.groupby('recording_id').mean().reset_index()
```

```python
import pandas as pd
import numpy as np
```

```python
score_1 = pd.read_csv('../../data/model/model-1/fold-1/scores.csv')
score_2 = pd.read_csv('../../data/model/model-1/fold-2/scores.csv')
score_3 = pd.read_csv('../../data/model/model-1/fold-3/scores.csv')
score_4 = pd.read_csv('../../data/model/model-1/fold-4/scores.csv')
score_5 = pd.read_csv('../../data/model/model-1/fold-5/scores.csv')
```

```python
score_1['ID'] = score_1.index.map(lambda x : 'development-' + str(x+1))
score_2['ID'] = score_2.index.map(lambda x : 'development-' + str(x+1))
score_3['ID'] = score_3.index.map(lambda x : 'development-' + str(x+1))
score_4['ID'] = score_4.index.map(lambda x : 'development-' + str(x+1))
score_5['ID'] = score_5.index.map(lambda x : 'development-' + str(x+1))
```

```python
score = score_1.append(score_2).append(score_3).append(score_4).append(score_5)
```

```python
submit = score.groupby('ID').mean().reset_index()
submit.to_csv('../../data/model.csv', index=False)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
```

```python
data = pd.read_csv('../../data/raw/train.csv')
```

```python
data['subset'] = -1
```

```python
folds = StratifiedKFold(random_state=2017, shuffle=True, n_splits=10)
```

```python
split = -1
for train_idx, valid_idx in folds.split(data.textID, data.sentiment):
    split += 1
    data.loc[valid_idx, 'subset'] = split 
```

```python
data.groupby(['subset','sentiment'])['textID'].count()
```

```python
data.to_csv('../../data/process/train.csv', index=False)
```

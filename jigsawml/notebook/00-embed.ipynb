```python
import pandas as pd
import numpy as np
import tensorflow as tf
import tensorflow_text
import tensorflow_hub as hub
```

```python
gpu = tf.config.experimental.list_physical_devices('GPU')[0]
tf.config.experimental.set_memory_growth(gpu, True)
```

```python
USE = hub.load("https://tfhub.dev/google/universal-sentence-encoder-multilingual-large/3")
```

```python
def fetchEmbed(text):
    return USE([text]).numpy()
```

```python
data = pd.read_csv('../../data/process/english/train_english.csv')
embeddings = data['comment_text'].map(fetchEmbed).values.tolist()
embeddings = np.vstack(embeddings).squeeze().round(5).astype(np.float32)
embeddings = pd.DataFrame(embeddings)
embeddings.columns = ['use_' + str(x) for x in range(512)]
data = data.join(embeddings)
data.to_csv('../../data/process/english/train_english_embed.csv', index=False)
```

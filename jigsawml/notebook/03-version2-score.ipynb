```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.optim.lr_scheduler import LambdaLR, ReduceLROnPlateau
from transformers.optimization import AdamW
from source.version2.data import scoreLoader
from source.version2.model import XLMModel
from source.version2.score import scoreModel
from multiprocessing import Pool
```

```python
def score(version):
    torch.cuda.empty_cache()
    score = scoreLoader()
    model = XLMModel()
    path = '../../model/version2/model_{}.pt'.format(version)
    weights = torch.load(path, map_location='cpu')
    weights = weights['model_state_dict']
    model.load_state_dict(weights)
    model = model.to('cuda:0')
    optimizer = AdamW(model.parameters(), lr=1e-6, weight_decay=0.)
    schedular = ReduceLROnPlateau(optimizer, factor=0.5, min_lr=1e-7)
    scorer = {}
    scorer['model'] = model
    scorer['data'] = score
    scorer['save'] = '../../model/version2/score_{}.csv'.format(version)
    scoreModel(**scorer)
    model = model.cpu()
    del model
    torch.cuda.empty_cache()
    return None
```

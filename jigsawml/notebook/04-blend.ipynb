```python
import pandas as pd
import numpy as np
from functools import reduce
```

```python
def postProcess(version):
    data0 = pd.read_csv(f'../../model/version{version}/score_0.csv')
    data1 = pd.read_csv(f'../../model/version{version}/score_1.csv')
    data2 = pd.read_csv(f'../../model/version{version}/score_2.csv')
    data3 = pd.read_csv(f'../../model/version{version}/score_3.csv')
    data4 = pd.read_csv(f'../../model/version{version}/score_4.csv')
    data0 = data0.append(data0.iloc[:63812,:].copy())
    data1 = data1.append(data1.iloc[:63812,:].copy())
    data2 = data2.append(data2.iloc[:63812,:].copy())
    data3 = data3.append(data3.iloc[:63812,:].copy())
    data4 = data4.append(data4.iloc[:63812,:].copy())
    print('Data:', data0.shape, data1.shape, data2.shape, data3.shape, data4.shape)
    combined = data0.append(data1).append(data2).append(data3).append(data4)
    combined = combined.groupby('id').mean().reset_index()
    combined.to_csv(f'../../model/version_{version}.csv', index=False)
    return combined
```

```python
version1 = postProcess('1') 
```

```python
version2 = postProcess('2') 
```

```python
combined = version1.append(version2).append(version2).append(version2)
```

```python
combined = combined.groupby('id').mean().reset_index()
```

```python
combined.to_csv('../../model/combined.csv', index=False)
```

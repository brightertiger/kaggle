```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from collections import Counter
```

### 2015 - Train

```python
data = pd.read_csv('../../data/pretrain/trainLabels15.csv')
data.columns = ['id_code','diagnosis']
data['fold'] = 0
data.diagnosis.value_counts()
```

```python
fold = StratifiedKFold(n_splits=10, random_state=2017)
counter = 0 
for train_idx, valid_idx in fold.split(data.index, data.diagnosis):
    counter += 1
    data.loc[valid_idx, 'fold'] = counter
```

```python
data.shape
```

```python
data.to_csv('../../data/pretrain/train_folds.csv', index=False)
```

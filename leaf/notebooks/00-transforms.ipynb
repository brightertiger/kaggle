```python
import cv2
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure
import albumentations as A
from albumentations.pytorch.transforms import ToTensorV2
%matplotlib inline
```

```python
def trainTF():
    transforms = []
    transforms  += [A.PadIfNeeded(600,800, p=1.0)]
    transforms += [A.Resize(600, 800, p=1.0)]
    transforms += [A.CenterCrop(512, 512, p=1.)]
    transforms += [A.Resize(512,512)]
    transforms = A.Compose(transforms, p=1.)
    return transforms
```

```python
def plotImage(image):
    tfms = trainTF()
    image = '../../data/raw/train_images/' + image
    image = cv2.imread(image, cv2.IMREAD_COLOR)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    f, ax = plt.subplots(1,2)
    ax[0].imshow(image)
    updt = tfms(image=image)['image']
    ax[1].imshow(updt)
    return None
```

```python
data = pd.read_csv('../../data/raw/data.csv')
```

```python
image = data.sample(1).iat[0,0]
plotImage(image)
```

```python
%autoreload
from processing.metric.pair import data_loader, score_loader
from modeling.metric.model import EmbedNet, Correct
from torch.nn import BCELoss
from torch.optim import Adam
from utility.optimizer import AdamW
from utility.schedular import CosineLR
from utility.checkpoint import save_model, load_model
from modeling.metric.pair import train_model, score_model
```

```python
def execute(train, valid, freeze, resnet, lr, load, save):
    name = 'pair'
    model = EmbedNet('../../model/metric/experiment-4/model.pth', freeze, resnet).float().cuda()
    optimizer = AdamW(model.parameters(), lr=lr, weight_decay=1)
    schedular = CosineLR(optimizer, T_max=100, T_mult=0.98, eta_min=1e-5)
    loss_fn = BCELoss()
    metric_fn = Correct()
    load_path = load
    save_path = save
    epochs = 1
    batch = 16
    arguments = []
    arguments += [name]
    arguments += [model]
    arguments += [train]
    arguments += [valid]
    arguments += [loss_fn]
    arguments += [metric_fn]
    arguments += [optimizer]
    arguments += [schedular]
    arguments += [save_path]
    arguments += [load_path]
    arguments += [epochs]
    arguments += [batch]
    train_model(*arguments)
    return None
```
